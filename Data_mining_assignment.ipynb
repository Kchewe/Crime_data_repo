```python
import pandas as pd

minority_class = 'Worms'
min_class_count = 130
class_counts = training_data['attack_cat'].value_counts()
min_class_count = class_counts.min()  # Get the count of the minority class

worms_data = training_data[training_data['attack_cat'] == minority_class]
resampled_data = [worms_data]

# Loop through each class and sample
for attack_cat in training_data['attack_cat'].unique():
    if attack_cat == minority_class:
        continue  # Skip the "Worms" class as it's already included
    class_data = training_data[training_data['attack_cat'] == attack_cat]
    class_resampled = class_data.sample(n=min_class_count, replace=False, random_state=42)
    resampled_data.append(class_resampled)

balanced_training_data = pd.concat(resampled_data, ignore_index=True)

```

```python
x_train = balanced_training_data.drop(["attack_cat"], axis=1)
y_train = balanced_training_data["attack_cat"]


x_test1 = test_data1.drop(["attack_cat"], axis=1)
y_test1 = test_data1["attack_cat"]

x_test2 = test_data2.drop(["attack_cat"], axis=1)
y_test2 = test_data2["attack_cat"]
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
categorical_columns_x = x_train.select_dtypes(include=['object']).columns
x_train = pd.get_dummies(x_train, columns=categorical_columns_x, drop_first = False)

y_train = le.fit_transform(y_train)

x_test1 = pd.get_dummies(x_test1, columns=categorical_columns_x, drop_first = False)
y_test1 = le.fit_transform(y_test1)
x_test2 = pd.get_dummies(x_test2, columns=categorical_columns_x, drop_first = False)
y_test2 = le.fit_transform(y_test2)
```

```python
import arff

# Create ARFF files for training and test sets
training_processed = pd.concat([x_train, pd.Series(y_train, name='attack_cat')], axis=1)
test1_processed = pd.concat([x_test1, pd.Series(y_test1, name='attack_cat')], axis=1)
test2_processed = pd.concat([x_test2, pd.Series(y_test2, name='attack_cat')], axis=1)

# Function to create ARFF file
def create_arff_file(df, filename):
    attributes = []
    for column in df.columns:
        if df[column].dtype == 'float64' or df[column].dtype == 'int64':
            attributes.append((column, 'REAL'))  # Numeric attributes
        elif column == 'attack_cat':
            unique_classes = df[column].unique()
            attributes.append((column, list(unique_classes)))  # Class labels
        else:
            attributes.append((column, ['0', '1']))  # Categorical (one-hot encoded)

    data_list = df.values.tolist()

    arff_data = {
        'description': f'{filename} dataset',
        'relation': filename,
        'attributes': attributes,
        'data': data_list
    }

    with open(f'{filename}.arff', 'w') as f:
        arff.dump(arff_data, f)

    print(f"ARFF file created: {filename}.arff")

# Create ARFF files for training and test sets
create_arff_file(training_processed, 'processed_training_data')
create_arff_file(test1_processed, 'processed_test_data1')
create_arff_file(test2_processed, 'processed_test_data2')
```
